```python
import numpy as np
import matplotlib.pyplot as plt

beta_arr=[0,0.5,0.9,0.99]
color=['r','b','g','y']

np.random.seed(42)

new_val=np.random.uniform(-12,12)
m_val=np.random.rand()

for i in range(len(beta_arr)):
  x=new_val
  lr=0.01
  epochs=30
  loss_history=[]
  m=m_val

  for epoch in range(epochs):
    loss_history.append(np.mean((x)**2))
    grad=2*x
    m=beta_arr[i]*x+grad
    x-=lr*m

  plt.plot(loss_history,range(epochs),color=color[i],label=f"beta value={beta_arr[i]}")

plt.title("Convergence vs beta values")
plt.xlabel("iterations")
plt.ylabel("loss")
plt.legend()
plt.show()
```
