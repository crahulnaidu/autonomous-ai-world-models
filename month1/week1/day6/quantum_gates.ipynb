```python
from qiskit import Quantum Circuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_vector

```

```python
# 1. Create a Quantum Circuit with 4 qubits and 4 classical bits
qc = QuantumCircuit(4, 4)

# 2. Apply Single-Qubit Gates
# Apply Pauli-X (NOT) gate to qubit 0
qc.x(0)

# Apply Pauli-Y gate to qubit 1
qc.y(1)

# Apply Pauli-Z gate to qubit 2
qc.z(2)

# Apply Hadamard (H) gate to qubit 3
qc.h(3)

# 3. Add Measurements (Maps qubit states to classical bits)
qc.measure([0, 1, 2, 3], [0, 1, 2, 3])

# 4. Display the Circuit
print(qc.draw(output='text'))
```
