```python
import numpy as np


def sigmoid(x):
    return np.where(x>=0,1.-0/(1+np.exp(-x)),np.exp(x)/(1+np.exp(x)))

def sigmoid_derivative(x):
    s=sigmoid(x)
    return s*(1.0-s)

def tanh(x):
    return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))

def tanh_derivative(x):
    t=tanh(x)
    return 1-t**2

def ReLU(x):
    return np.maximum(0.0,x)

def ReLU_derivative(x):
    return (x>0.0).astype(float)

def LeakyReLU(x,alpha=0.01):
    return np.where(x>0.0,x,alpha*x)

def LeakyReLU_derivative(x,alpha=0.01):
    return np.where(x>0.0,1.0,alpha)
```

```python
plt.title("Their derivative")

plt.plot(x,sigmoid_derivative(x),color='blue',label="sigmoid derivative")
plt.plot(x,tanh_derivative(x),color='red',label="tanh derivative")
plt.plot(x,ReLU_derivative(x),color='yellow',label="ReLU derivative")
plt.plot(x,LeakyReLU_derivative(x),color='black',label="leakyReLU derivative")
plt.legend()
```
